```python
import tensorflow as tf
import os
```

```python
review_ds = tf.data.Dataset.list_files('Reviews/*/*', shuffle=False)
```

```python
def extract_review_and_labels(file_path):
  return tf.io.read_file(file_path), tf.strings.split(file_path, os.path.sep)[-2]
```

```python
review_ds = review_ds.map(extract_review_and_labels)
for review, label in review_ds:
  print('Reviews:', review.numpy()[:50])
  print('Label:', label.numpy())
```

```python
review_ds = review_ds.filter(lambda review, label: review!= '')
for review, label in review_ds:
  print('Reviews:', review.numpy()[:50])
  print('Label:', label.numpy())
```
